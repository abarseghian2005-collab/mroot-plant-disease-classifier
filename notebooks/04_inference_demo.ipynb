# Try the final model locally

The public Space is the easiest demo. This notebook documents the local path.
It downloads about 691 MB of verified weights only when explicitly enabled.
The model has 39 disease labels and no healthy category.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
REPORTS = ROOT / "reports/raw"
plt.style.use("default")
plt.rcParams.update({"axes.spines.top": False, "axes.spines.right": False})
```

```python
RUN_INFERENCE = False
IMAGE_PATH = Path("leaf.jpg")

if RUN_INFERENCE:
    import sys
    from PIL import Image
    sys.path.insert(0, str(ROOT))
    from src.mroot.assets import ensure_model_assets
    from src.mroot.inference import MrootEnsemble
    model = MrootEnsemble(ensure_model_assets(), device="cpu")
    with Image.open(IMAGE_PATH) as source:
        scores = model.predict(source.convert("RGB"))
    print(sorted(scores.items(), key=lambda item: item[1], reverse=True)[:5])
else:
    print("No models downloaded and no image evaluated. Set RUN_INFERENCE when ready.")
```
